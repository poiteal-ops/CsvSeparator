# Convert a CSV file

Use this notebook if you have a CSV file with commas inside text fields (addresses, descriptions, ...) and want to fix it, without using a terminal.

It runs the exact same command-line tool described in the [README](../README.md) — it doesn't reimplement any logic, it just wraps `python -m src.main` so you can fill in a file path and run cells instead of typing flags.

For debugging/exploring *how* the detection and repair logic works internally, see `explore_csvseparator.ipynb` instead. This notebook is for the practical "I have a file, fix it" case.

**Prerequisite:** the first row of your file must be a header row (column names). Every check below assumes that.

## 1. Point this at your file

Set `file_path` below. It can be:
- a path relative to the project root (e.g. `"data/input/my_export.csv"`), or
- an absolute path anywhere on disk (e.g. `"C:/Users/you/Downloads/export.csv"`).

```python
file_path = "data/input/sample_complex.csv"  # <-- change this
```

## 2. Check for issues (nothing is written yet)

This only analyzes the file and reports what it finds. Exit code `0` means clean, `2` means it found rows with likely comma problems — either way, no file is written by this step.

```python
!python -m src.main --input "{file_path}" --analyze-only
```

## 3. Convert it

Pick a `separator` (`pipe`, `semicolon`, `tab`, or `comma`) and run the cell below. `--yes` is required to actually write a file — without it, the tool only reports what it *would* do.

The converted file is written to `data/output/` (same base name, with the separator in the filename), alongside an `.issues.json` report if any rows were flagged.

```python
separator = "pipe"  # pipe | semicolon | tab | comma

!python -m src.main --input "{file_path}" --separator {separator} --yes
```

## Optional: repair broken rows instead of just flagging them

By default, a row with extra columns (from an unescaped comma) is only *flagged* in the output, not fixed. Add `--quote-fix` to attempt an actual repair: it finds the one field that likely absorbed the stray comma(s) and re-quotes it.

It only repairs a row when exactly one field can be identified with confidence — ambiguous rows are left flagged, unrepaired, same as without this flag. It works with any separator, including keeping `comma` (fix the file in place without switching separators).

```python
!python -m src.main --input "{file_path}" --separator {separator} --quote-fix --yes
```
